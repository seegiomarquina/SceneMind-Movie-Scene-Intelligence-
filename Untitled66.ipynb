```python
# ============================================================
# DAY 14 — SCENEMIND AI
# Movie Scene Intelligence & Visual Similarity Detection
#
# Computer Vision | ResNet50 | Cosine Similarity
# K-Means | PCA | DSA Heap | GitHub Exports
# ============================================================

import os
import sys
import cv2
import json
import time
import heapq
import zipfile
import requests
import warnings
import subprocess
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from pathlib import Path
from PIL import Image
from IPython.display import display

from sklearn.metrics.pairwise import cosine_similarity
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import normalize

warnings.filterwarnings("ignore")

# ------------------------------------------------------------
# 1. PROJECT CONFIGURATION
# ------------------------------------------------------------

PROJECT = Path("/content/SceneMind_AI")

FRAMES_DIR = PROJECT / "frames"
OUTPUT_DIR = PROJECT / "outputs"
PLOTS_DIR = PROJECT / "visualizations"

for folder in [FRAMES_DIR, OUTPUT_DIR, PLOTS_DIR]:
    folder.mkdir(parents=True, exist_ok=True)

# Updated to a highly reliable public CDN repository video that does not block Colab
VIDEO_URL = (
    "https://raw.githubusercontent.com/intel-iot-devkit/"
    "sample-videos/master/person-bicycle-car-detection.mp4"
)

VIDEO_PATH = PROJECT / "sample_video.mp4"

FRAME_INTERVAL = 2
MAX_FRAMES = 50
IMAGE_SIZE = (224, 224)

print("=" * 65)
print("              SCENEMIND AI INITIALIZING")
print("=" * 65)

# ------------------------------------------------------------
# 2. DOWNLOAD SAMPLE VIDEO
# ------------------------------------------------------------

if not VIDEO_PATH.exists():
    print("\nDownloading public sample video...")

    # Add standard browser User-Agent headers to avoid standard server blocks
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36"
    }

    response = requests.get(
        VIDEO_URL,
        headers=headers,
        stream=True,
        timeout=120
    )

    response.raise_for_status()

    with open(VIDEO_PATH, "wb") as f:
        for chunk in response.iter_content(chunk_size=1024*1024):
            if chunk:
                f.write(chunk)

    print("Video downloaded successfully!")

else:
    print("Existing video found.")

# ------------------------------------------------------------
# 3. VIDEO INFORMATION
# ------------------------------------------------------------

cap = cv2.VideoCapture(str(VIDEO_PATH))

if not cap.isOpened():
    raise RuntimeError(
        "Unable to open video. Try uploading another MP4 video."
    )

fps = cap.get(cv2.CAP_PROP_FPS)
total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

duration = total_frames / fps if fps > 0 else 0

print("\nVIDEO INFORMATION")
print("-" * 45)
print("Resolution:", width, "x", height)
print("FPS:", fps)
print("Total frames:", total_frames)
print("Duration:", round(duration, 2), "seconds")

cap.release()

# ------------------------------------------------------------
# 4. EXTRACT FRAMES USING OPENCV
# ------------------------------------------------------------

print("\nExtracting video frames...")

cap = cv2.VideoCapture(str(VIDEO_PATH))

frame_records = []
frame_count = 0
frame_number = 0

while True:

    success, frame = cap.read()

    if not success:
        break

    timestamp = (
        frame_number / fps if fps > 0 else 0
    )

    if timestamp >= frame_count * FRAME_INTERVAL:

        if frame_count >= MAX_FRAMES:
            break

        frame = cv2.resize(frame, IMAGE_SIZE)

        filename = f"scene_{frame_count:04d}.jpg"

        filepath = FRAMES_DIR / filename

        cv2.imwrite(str(filepath), frame)

        frame_records.append({
            "scene_id": frame_count,
            "timestamp_seconds": round(timestamp, 2),
            "frame_path": str(filepath),
            "filename": filename
        })

        frame_count += 1

    frame_number += 1

cap.release()

frames_df = pd.DataFrame(frame_records)

if len(frames_df) < 3:
    raise RuntimeError(
        "Not enough frames extracted. Try another video."
    )

print("Frames extracted:", len(frames_df))

display(frames_df.head())

# ------------------------------------------------------------
# 5. VISUALIZE EXTRACTED FRAMES
# ------------------------------------------------------------

print("\nDISPLAYING SAMPLE FRAMES")

sample_count = min(6, len(frames_df))

fig, axes = plt.subplots(
    2, 3,
    figsize=(15, 8)
)

for ax, (_, row) in zip(
    axes.flatten(),
    frames_df.head(sample_count).iterrows()
):

    image = cv2.imread(row["frame_path"])
    image = cv2.cvtColor(
        image,
        cv2.COLOR_BGR2RGB
    )

    ax.imshow(image)
    ax.set_title(
        f"Scene {row['scene_id']} | "
        f"{row['timestamp_seconds']} sec"
    )
    ax.axis("off")

for ax in axes.flatten()[sample_count:]:
    ax.axis("off")

plt.suptitle(
    "SceneMind AI — Extracted Frames",
    fontsize=16
)

plt.tight_layout()

plt.savefig(
    PLOTS_DIR / "extracted_frames.png",
    dpi=150
)

plt.show()

# ------------------------------------------------------------
# 6. LOAD PRETRAINED RESNET50
# ------------------------------------------------------------

print("\nLoading pretrained ResNet50...")

import tensorflow as tf

from tensorflow.keras.applications.resnet50 import (
    ResNet50,
    preprocess_input
)

model = ResNet50(
    weights="imagenet",
    include_top=False,
    pooling="avg",
    input_shape=(224, 224, 3)
)

print("ResNet50 loaded successfully!")

# ------------------------------------------------------------
# 7. EXTRACT VISUAL EMBEDDINGS
# ------------------------------------------------------------

print("\nGenerating visual embeddings...")

image_batch = []

for path in frames_df["frame_path"]:

    image = Image.open(path).convert("RGB")

    image = image.resize(IMAGE_SIZE)

    image = np.array(image)

    image_batch.append(image)

image_batch = np.array(
    image_batch,
    dtype=np.float32
)

processed_batch = preprocess_input(
    image_batch.copy()
)

embeddings = model.predict(
    processed_batch,
    batch_size=16,
    verbose=1
)

embeddings = normalize(embeddings)

print("\nEMBEDDING INFORMATION")
print("-" * 45)
print("Number of scenes:", embeddings.shape[0])
print("Embedding dimensions:", embeddings.shape[1])

np.save(
    OUTPUT_DIR / "scene_embeddings.npy",
    embeddings
)

# ------------------------------------------------------------
# 8. BUILD SCENE SIMILARITY ENGINE
# ------------------------------------------------------------

print("\nBuilding similarity engine...")

similarity_matrix = cosine_similarity(
    embeddings
)

print(
    "Similarity matrix shape:",
    similarity_matrix.shape
)

# ------------------------------------------------------------
# 9. DSA — TOP K SIMILAR SCENES
# ------------------------------------------------------------

print("\nDSA IMPLEMENTATION")
print("=" * 55)

def top_k_similar_scenes(
    similarity_scores,
    query_index,
    k=5
):

    """
    Return top K similar scenes using a heap.

    Excludes the query scene itself.

    Time complexity:
        O(N log K) for heap ranking,
        after similarity scores have been calculated.

    Space complexity:
        O(K) for the selected heap/result.
    """

    candidates = [
        (float(score), index)
        for index, score in enumerate(
            similarity_scores
        )
        if index != query_index
    ]

    top_results = heapq.nlargest(
        min(k, len(candidates)),
        candidates,
        key=lambda item: (item[0], -item[1])
    )

    return top_results


# ------------------------------------------------------------
# 10. INTERACTIVE SCENE SEARCH FUNCTION
# ------------------------------------------------------------

def search_similar_scenes(
    query_index,
    top_k=5
):

    if query_index < 0 or query_index >= len(embeddings):
        raise ValueError("Invalid scene index.")

    results = top_k_similar_scenes(
        similarity_matrix[query_index],
        query_index,
        top_k
    )

    output = []

    print("\nQUERY SCENE:", query_index)

    query_image = cv2.imread(
        frames_df.iloc[query_index]["frame_path"]
    )

    query_image = cv2.cvtColor(
        query_image,
        cv2.COLOR_BGR2RGB
    )

    plt.figure(figsize=(6, 4))
    plt.imshow(query_image)
    plt.title(f"Query Scene {query_index}")
    plt.axis("off")
    plt.show()

    print("\nTOP MATCHING SCENES")
    print("-" * 55)

    for rank, (score, index) in enumerate(
        results,
        start=1
    ):

        row = frames_df.iloc[index]

        output.append({
            "rank": rank,
            "query_scene": query_index,
            "matched_scene": index,
            "timestamp_seconds": row[
                "timestamp_seconds"
            ],
            "similarity_score": round(score, 5),
            "filename": row["filename"]
        })

        print(
            f"Rank {rank}: Scene {index} | "
            f"Similarity: {score:.4f} | "
            f"Time: {row['timestamp_seconds']} sec"
        )

    return pd.DataFrame(output)


# TEST THE SEARCH ENGINE

recommendations = search_similar_scenes(
    query_index=0,
    top_k=5
)

display(recommendations)

# ------------------------------------------------------------
# 11. VISUALIZE SIMILARITY RESULTS
# ------------------------------------------------------------

fig, axes = plt.subplots(
    1,
    min(6, len(recommendations) + 1),
    figsize=(18, 5)
)

if not isinstance(axes, np.ndarray):
    axes = np.array([axes])

query_path = frames_df.iloc[0]["frame_path"]

query_image = cv2.cvtColor(
    cv2.imread(query_path),
    cv2.COLOR_BGR2RGB
)

axes[0].imshow(query_image)
axes[0].set_title("Query Scene")
axes[0].axis("off")

for ax, (_, row) in zip(
    axes[1:],
    recommendations.iterrows()
):

    image = cv2.imread(
        frames_df.iloc[
            int(row["matched_scene"])
        ]["frame_path"]
    )

    image = cv2.cvtColor(
        image,
        cv2.COLOR_BGR2RGB
    )

    ax.imshow(image)

    ax.set_title(
        f"Match {row['rank']}\n"
        f"Score: {row['similarity_score']}"
    )

    ax.axis("off")

plt.suptitle("Visual Scene Similarity Search")

plt.tight_layout()

plt.savefig(
    PLOTS_DIR / "similarity_results.png",
    dpi=150
)

plt.show()

# ------------------------------------------------------------
# 12. K-MEANS SCENE CLUSTERING
# ------------------------------------------------------------

print("\nCLUSTERING SCENES")

n_clusters = min(
    5,
    len(embeddings) - 1
)

kmeans = KMeans(
    n_clusters=n_clusters,
    random_state=42,
    n_init=10
)

cluster_labels = kmeans.fit_predict(
    embeddings
)

frames_df["cluster"] = cluster_labels

# ------------------------------------------------------------
# 13. CLUSTER PROFILES
# ------------------------------------------------------------

cluster_profiles = (
    frames_df.groupby("cluster")
    .agg(
        total_scenes=("scene_id", "count"),
        first_timestamp=("timestamp_seconds", "min"),
        last_timestamp=("timestamp_seconds", "max")
    )
    .reset_index()
)

print("\nCLUSTER PROFILES")

display(cluster_profiles)

cluster_profiles.to_csv(
    OUTPUT_DIR / "cluster_profiles.csv",
    index=False
)

# ------------------------------------------------------------
# 14. EVALUATE CLUSTERING
# ------------------------------------------------------------

silhouette = None

if 2 <= len(set(cluster_labels)) < len(embeddings):

    silhouette = silhouette_score(
        embeddings,
        cluster_labels
    )

    print(
        "Silhouette Score:",
        round(silhouette, 4)
    )

else:
    print("Silhouette score unavailable.")

# ------------------------------------------------------------
# 15. PCA — DIMENSIONALITY REDUCTION
# ------------------------------------------------------------

print("\nRUNNING PCA...")

pca = PCA(
    n_components=2,
    random_state=42
)

pca_embeddings = pca.fit_transform(
    embeddings
)

pca_df = pd.DataFrame({
    "PCA_1": pca_embeddings[:, 0],
    "PCA_2": pca_embeddings[:, 1],
    "cluster": cluster_labels,
    "scene_id": frames_df["scene_id"]
})

plt.figure(figsize=(11, 7))

scatter = plt.scatter(
    pca_df["PCA_1"],
    pca_df["PCA_2"],
    c=pca_df["cluster"],
    cmap="viridis",
    s=80,
    alpha=0.8
)

plt.colorbar(
    scatter,
    label="Cluster"
)

plt.title(
    "SceneMind AI — PCA Scene Clusters"
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")

plt.grid(alpha=0.2)

plt.tight_layout()

plt.savefig(
    PLOTS_DIR / "pca_clusters.png",
    dpi=150
)

plt.show()

print(
    "PCA explained variance:",
    round(
        sum(pca.explained_variance_ratio_),
        4
    )
)

# ------------------------------------------------------------
# 16. CLUSTER DISTRIBUTION
# ------------------------------------------------------------

cluster_counts = frames_df[
    "cluster"
].value_counts().sort_index()

plt.figure(figsize=(9, 5))

plt.bar(
    cluster_counts.index.astype(str),
    cluster_counts.values
)

plt.title("Vehicle/Scene Visual Cluster Distribution")

plt.xlabel("Cluster")
plt.ylabel("Number of Frames")

plt.tight_layout()

plt.savefig(
    PLOTS_DIR / "cluster_distribution.png",
    dpi=150
)

plt.show()

# ------------------------------------------------------------
# 17. SCENE EXPLORER
# ------------------------------------------------------------

def explore_cluster(cluster_id):

    selected = frames_df[
        frames_df["cluster"] == cluster_id
    ]

    if selected.empty:
        print("No scenes found.")
        return

    print(
        f"\nCluster {cluster_id}: "
        f"{len(selected)} scenes"
    )

    count = min(8, len(selected))

    fig, axes = plt.subplots(
        2, 4,
        figsize=(16, 8)
    )

    for ax, (_, row) in zip(
        axes.flatten(),
        selected.head(count).iterrows()
    ):

        image = cv2.cvtColor(
            cv2.imread(row["frame_path"]),
            cv2.COLOR_BGR2RGB
        )

        ax.imshow(image)

        ax.set_title(
            f"Scene {row['scene_id']}"
        )

        ax.axis("off")

    for ax in axes.flatten()[count:]:
        ax.axis("off")

    plt.tight_layout()
    plt.show()


# Example
explore_cluster(0)

# ------------------------------------------------------------
# 18. EXPORT ALL SCENE DATA
# ------------------------------------------------------------

print("\nEXPORTING PROJECT OUTPUTS...")

frames_df.to_csv(
    OUTPUT_DIR / "scene_index.csv",
    index=False
)

recommendations.to_csv(
    OUTPUT_DIR / "sample_recommendations.csv",
    index=False
)

pca_df.to_csv(
    OUTPUT_DIR / "pca_coordinates.csv",
    index=False
)

np.save(
    OUTPUT_DIR / "cluster_labels.npy",
    cluster_labels
)

# Save experiment information

metrics = {
    "project": "SceneMind AI",
    "video": VIDEO_PATH.name,
    "frames_extracted": int(len(frames_df)),
    "embedding_dimensions": int(embeddings.shape[1]),
    "number_of_clusters": int(n_clusters),
    "silhouette_score": (
        float(silhouette)
        if silhouette is not None
        else None
    ),
    "pca_explained_variance": float(
        sum(pca.explained_variance_ratio_)
    ),
    "dsa_problem": "Top K Similar Elements",
    "dsa_method": "Heap / Priority Queue"
}

with open(
    OUTPUT_DIR / "metrics.json",
    "w"
) as f:

    json.dump(
        metrics,
        f,
        indent=4
    )

# ------------------------------------------------------------
# 19. REQUIREMENTS FILE
# ------------------------------------------------------------

requirements = """
numpy
pandas
opencv-python
Pillow
matplotlib
scikit-learn
tensorflow
requests
"""

with open(
    PROJECT / "requirements.txt",
    "w"
) as f:

    f.write(requirements.strip())

# ------------------------------------------------------------
# 20. CREATE GITHUB PACKAGE
# ------------------------------------------------------------

ZIP_PATH = Path(
    "/content/SceneMind_AI_GitHub_Package.zip"
)

with zipfile.ZipFile(
    ZIP_PATH,
    "w",
    zipfile.ZIP_DEFLATED
) as zipf:

    for file in PROJECT.rglob("*"):

        if file.is_file():

            zipf.write(
                file,
                file.relative_to(PROJECT.parent)
            )

# ------------------------------------------------------------
# 21. FINAL PROJECT REPORT
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("             SCENEMIND AI — COMPLETED")
print("=" * 65)

print("\nPROJECT SUMMARY")
print("-" * 45)

print("Frames processed:", len(frames_df))
print("Feature dimensions:", embeddings.shape[1])
print("Clusters created:", n_clusters)

if silhouette is not None:
    print("Silhouette Score:", round(silhouette, 4))

print(
    "PCA explained variance:",
    round(
        sum(pca.explained_variance_ratio_),
        4
    )
)

print("\nPROJECT FILES")
print("-" * 45)

print("Project folder:", PROJECT)
print("Outputs:", OUTPUT_DIR)
print("Visualizations:", PLOTS_DIR)
print("GitHub package:", ZIP_PATH)

print("\nSCENEMIND AI IS READY FOR GITHUB PREPARATION!")

# Download ZIP from Google Colab

from google.colab import files

files.download(str(ZIP_PATH))
```
